### 1. Визначаємо параметри задачі

```python
sample_mean = data.mean()
population_mean = 20
population_std = np.std(data, ddof=1)
sample_size = len(data)
alpha = 0.05
```

```python
print(f'Середнє значення в вибірці x̄ = {sample_mean:.2f}.')
print(f'Середнє значення в популяції μ₀ = {population_mean}.')
print(f'Cтандартне відхилення в популяції σ = {population_std:.2f}.')
print(f'Розмір вибірки n = {sample_size}.')
print(f'Рівень значущості α = {alpha}.')
```

### 2. Формулюємо гіпотези

**Гіпотези**   

H₀: μ = 20 (середній чек не підвищився)  
H₁: μ > 20 (середній чек підвищився)

Тип тесту: правосторонній (one-tailed)  

### 3. Обчислюємо стандартну помилку


```python
standard_error = population_std / np.sqrt(sample_size)
```

```python
print(f'Обрахуємо стандартну помилку за формулою: SE = σ / √n.')
print(f'Стандартна помилка SE = {standard_error:.3f}.')
print(f"""Це означає, що вибіркове середнє відхиляється
від справжнього середнього на ± {standard_error:.3f}.""")
```

### 4. Розраховуємо z-статистику

```python
z_statistic = (sample_mean - population_mean) / standard_error
```

```python
print(f'Обрахуємо z-статистику за формулою: z = (x̄ - μ₀) / SE.')
print(f'Після обрахунків отримаємо z = {z_statistic:.3f}.')
print(f"""Це означає, що вибіркове середнє відхиляється на {z_statistic:.3f}
стандартних помилок від середнього популяції.""")
```

### 5. Знаходимо критичне значення

```python
z_critical = stats.norm.ppf(1 - alpha)
```

```python
print(f"""Шукаємо z-критичне для рівня значущості α = {alpha}
та правостороннього тесту: z = {z_critical:.3f}.""")
```

### 6. Обчислюємо p-value

```python
p_value = 1 - stats.norm.cdf(z_statistic)
```

```python
print(f"Шукаємо p-value для правостороннього тесту: p-value = {p_value:.2f}.")
```

### 7. Приймаємо рішення

**Метод 1: Порівняння z-статистики з критичним значенням**

```python
print(f"z-статистика = {z_statistic:.3f}.")
print(f"z-критичне = {z_critical:.3f}.")
print(f"Порівняння: {z_statistic:.3f} {'>' if z_statistic > z_critical else '≤'} {z_critical:.3f}.")

if z_statistic > z_critical:
  print ('z-статистика більше за z- критичне. Відхиляємо гіпотезу H₀.')
else:
  print('z-статистика не більше за z- критичне. Не відхиляємо гіпотезу H₀.')
```

**Метод 2: Порівняння p-value з рівнем значущості**

```python
print(f'p-value = {p_value:.2f}.')
print(f'Рівень значущості α = {alpha}.')
print(f'Порівняння {p_value:.2f} {'<' if p_value < alpha else '≥'} {alpha}.')

if p_value < alpha:
  print ('p-value менше за рівень значущості -> Відхиляємо гіпотезу H₀')
else:
  print('p-value не менше за рівень значущості -> Не відхиляємо гіпотезу H₀.')
```

```python
if (z_statistic > z_critical) & (p_value < alpha):
  print('Ура! Ми підвищили середній чек в липні 2025 року.')
else:
  print('На жаль, в липні 2025 року не вдалось підвищити середній чек.')
```

Чи зійшлись значення z-статистику та р-значення в цьому завданні з попередніми обчисленнями?

```python
from statsmodels.stats import weightstats as stests

z_stat_sm, p_val_sm = stests.ztest(data,
                                  value=population_mean,
                                  alternative='larger')
print('Результати z-тесту:')
print(f'z-статистика = {z_stat_sm:.3f}')
print(f'p-value = {p_val_sm:.2f}')

print('')
if p_val_sm < alpha:
  print(f'p-value = {p_val_sm:.2f} < α = {alpha} -> Відхиляємо гіпотезу H₀.')
  print("""Висновок: На рівні значущості 5% середній чек в липні
2025 року вдалось підвищити порівняно з минулими місяцями.""")
else:
  print(f'p-value = {p_val_sm:.2f} ≥ α = {alpha} -> Не відхиляємо гіпотезу H₀.')
  print("""Висновок: на жаль, не маємо достатньо підстав вважати, що
вдалось підвищити чек в липні 2025 року, порівняно з минулими місяцями.""")

print('')
print('Чи співпали розрахунки в двох методах?')
print('Z-статистика:')
if z_statistic == z_stat_sm:
  print(f'Розрахунки z-статистики співпали: {z_statistic:.3f} = {z_stat_sm:.3f}')
else:
  print(f'Розрахунки z-статистики НЕ співпали: {z_statistic:.3f} ≠ {z_stat_sm:.3f}')

print('')
print('p-value:')
if p_value == p_val_sm:
  print(f'Розрахунки p-value співпали: {p_value:.2f} = {p_val_sm:.2f}')
else:
  print(f'Розрахунки p-value НЕ співпали: {p_value:.32f} ≠ {p_val_sm:.2f}')
```
