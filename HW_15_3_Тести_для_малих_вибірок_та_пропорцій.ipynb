**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

### Маємо правосторонній тест

**Гіпотези:**  

H₀: μ = 123.7 (середній час до покупки не збільшився)  
H₁: μ > 123.7 (середній час до покупки збільшився)

```python
import numpy as np
from scipy import stats
```

```python
# Дані:
times = np.array([128, 135, 121, 142, 126, 151, 123])
mu0 = 123.7
alpha = 0.1
```

### Обчислимо вибіркову статистику: $\bar x$, $s$, $n$.

```python
x_bar = np.mean(times)
s = np.std(times, ddof=1)
n = len(times)
```

### Для правостороннього тесту обчислимо: t-статистику, t-критичне та p-value:

```python
t_stat = (x_bar - mu0) / (s / np.sqrt(n))
df = n - 1
t_crit = stats.t.ppf(1 - alpha, df)
p_value = 1 - stats.t.cdf(t_stat, df)
```

### Виводимо обчислення:

```python
print (f'Середнє вибірки: {x_bar:.3f}')
print (f'Стандартне відхилення: {s:.3f}')
print (f't-статистика: {t_stat:.3f}')
print (f't-критичне: {t_crit:.3f}')
print (f'p-value: {p_value:.3f}')
```

### Висновок на основі t-значень:

```python
if t_stat > t_crit:
  print('t-статистика більше за t-критичне. Відхиляємо гіпотезу H₀.')
else:
  print('t-статистика не більше за t-критичне. Не відхиляємо гіпотезу H₀.')
```

### Висновок на основі p-value:

```python
if p_value < alpha:
  print('p-value менше за рівень значущості -> Відхиляємо гіпотезу H₀')
else:
  print('p-value не менше за рівень значущості -> Не відхиляємо гіпотезу H₀')
```

```python
print (f"""На рівні значущості {alpha} відхиляємо гіпотезу H₀.
Середній час до покупки на сайті збільшився після редизайну.""")
```

### Чи зміниться ваше рішення при зміні рівня значущості на $\alpha = 0.05$?

Рішення не зміниться, ми так само відхиляємо гіпотезу H₀, але вже не так впевнено, так як p_value зовсім трохи менше за alpha -  0.042 < 0.05.

**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

```python
import statsmodels.api as sm
from statsmodels.stats.proportion import proportions_ztest, proportion_confint
```

```python
# Дані:
count = 311
nobs = 1200
p0 = 0.23
alpha = 0.03
```

```python
# z-тест для пропорції
stat, pval = proportions_ztest(count, nobs, p0, alternative='larger')
confint = proportion_confint(count, nobs, alpha, method='normal')
```

```python
print('Результати z-тесту для пропорції:')
print(f'Z-статистика = {stat:.3f}')
print(f'p_value = {pval:.4f}')
print(f'97% довірчий інтервал: {confint}')
```

### Інтерпретація результатів:

***1. Маємо правосторонній тест:***

 - H₀: p = 0.23

 - H₁: p > 0.23

 - Так як, p_value < alpha -> 0.0106 < 0.03 -> Відхиляємо H₀. Маємо достатньо підстав стверджувати, що більше ніж 23% дорослих впізнають логотип компанії.

***2. 97% довірчий інтервал (0.232, 0.287):***
 - З ймовірністю 97% справжня частка лежить у межах від 23.2% до 28.7%.
